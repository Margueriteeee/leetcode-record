# 376 摆动序列
摆动序列定义：连续数字时间的差严格在正数和负数之间交替，该数字序列称为摆动序列  
- 第一个差如果存在，正负都可以  
- 少于两个元素的序列也算作摆动序列

输入：一个数字序列  
输出：通过对原始序列进行删除（或者不删除）来获得子序列，输出作为摆动序列的最长子序列的长度。剩下的元素需要保持原始顺序   


# 思路
按照 **坡度** 的图形思维理解，摆动序列就是不断出现峰值而没有单调坡度  
  

目标：经过修改后的序列拥有最多的局部峰值（局部最优获得全局最优）  

题目要求输出的是摆动序列的最长长度，所以可以不删除元素，只统计局部峰值数量  

统计峰值的条件：（后-前）  
- 当前差和前一个差异号  
- 也就是 prediff < 0 && curdiff > 0 或者 prediff > 0 && curdiff < 0  
- 也可以写成 prediff * curdiff < 0  

根据坡度的 起伏情况 进行排除：  
- 上下坡中有平坡：把平坡多余的删除，只留下一个形成局部峰值 
  - 每次要么删除前面的三个：当（当前差<0 && 前一个差=0）时删除
  - 要么删除后面的三个
